```python

# Compact categorical summary for both datasets

def inspect_categories(df, columns):
    summary = []

    for col in columns:
        if col in df.columns:
            values = df[col].dropna().unique()

            summary.append({
                "Column": col,
                "Unique Values": df[col].nunique(dropna=True),
                "Missing": df[col].isna().sum(),
                "Categories": ", ".join(map(str, values[:12]))
            })

    return pd.DataFrame(summary)

print("TRANSACTION CATEGORICAL SUMMARY")
display(inspect_categories(transactions, transaction_categories))

print("ACCOUNT CATEGORICAL SUMMARY")
display(inspect_categories(accounts, account_categories))
```

```python


# STEP 2.4.5: Validate the amount-to-average ratio

# Select records with a non-zero spending baseline
valid_baseline = (
    transactions["avg_transaction_amount_30d"].notna()
    & (transactions["avg_transaction_amount_30d"] != 0)
)

# Calculate a candidate ratio without changing the dataset
calculated_ratio = (
    transactions.loc[valid_baseline, "amount"].abs()
    / transactions.loc[
        valid_baseline, "avg_transaction_amount_30d"
    ].abs()
)

reported_ratio = transactions.loc[
    valid_baseline, "amount_to_avg_ratio"
]

# Compare calculated and reported values
difference = (calculated_ratio - reported_ratio).abs()

print("Records checked:", valid_baseline.sum())

print(
    "Matching ratios (tolerance 0.01):",
    (difference <= 0.01).sum()
)

print(
    "Non-matching ratios:",
    (difference > 0.01).sum()
)

print(
    "Maximum absolute difference:",
    difference.max()
)

print(
    "\nZero spending baselines:",
    (transactions["avg_transaction_amount_30d"] == 0).sum()
)

# Inspect examples of mismatches
comparison = transactions.loc[
    valid_baseline,
    [
        "amount",
        "avg_transaction_amount_30d",
        "amount_to_avg_ratio"
    ]
].copy()

comparison["calculated_ratio"] = calculated_ratio
comparison["difference"] = difference

print("\nSAMPLE NON-MATCHING RATIOS")

display(
    comparison.loc[
        comparison["difference"] > 0.01
    ].head(10)
)
```

```python

# STEP 2.4.6: Test the minimum-denominator hypothesis

# Calculate the candidate formula
candidate_ratio = (
    transactions["amount"].abs()
    / transactions["avg_transaction_amount_30d"]
      .abs()
      .clip(lower=1)
)

reported_ratio = transactions["amount_to_avg_ratio"]

difference = (candidate_ratio - reported_ratio).abs()

print("Total transactions:", len(transactions))

print(
    "Matching ratios (tolerance 0.01):",
    (difference <= 0.01).sum()
)

print(
    "Non-matching ratios:",
    (difference > 0.01).sum()
)

print(
    "Maximum absolute difference:",
    difference.max()
)

print("\nSAMPLE NON-MATCHING RECORDS")

comparison = transactions[
    [
        "amount",
        "avg_transaction_amount_30d",
        "amount_to_avg_ratio"
    ]
].copy()

comparison["candidate_ratio"] = candidate_ratio
comparison["difference"] = difference

display(
    comparison.loc[
        comparison["difference"] > 0.01
    ].head(10)
)
```

### Findings — Amount-to-Average Ratio Validation

The investigation found that 125,857 out of 126,000 transactions (99.89%) matched the proposed ratio calculation within a tolerance of 0.01.

The calculation appears consistent with dividing the absolute transaction amount by the greater of the absolute 30-day spending baseline and 1.

The remaining 143 differences were small, with a maximum absolute difference of approximately 0.1831. Rounding or precision differences may explain these discrepancies, although this has not been confirmed.

**Decision:** Retain the original `amount_to_avg_ratio` values for now. Verify the feature's historical calculation before model development to prevent potential data leakage.

No data was modified during this validation.

```python

# STEP 2.4.13: Validate transaction velocity

import numpy as np
import pandas as pd

# Temporary inspection dataset
velocity_check = transactions[
    ["transaction_id", "account_id", "timestamp",
     "transaction_velocity_1h"]
].copy()

velocity_check["timestamp"] = pd.to_datetime(
    velocity_check["timestamp"], utc=True
)

# Sort chronologically within each account
velocity_check = velocity_check.sort_values(
    ["account_id", "timestamp", "transaction_id"]
)

# Count earlier transactions in the trailing 60 minutes
def calculate_velocity(group):
    times = group["timestamp"].astype("int64").to_numpy()
    window_ns = pd.Timedelta(hours=1).value

    left = np.searchsorted(
        times, times - window_ns, side="right"
    )

    right = np.searchsorted(
        times, times, side="left"
    )

    return pd.Series(
        right - left,
        index=group.index
    )

velocity_check["calculated_velocity"] = (
    velocity_check.groupby(
        "account_id", group_keys=False
    ).apply(calculate_velocity)
)

velocity_check["velocity_match"] = (
    velocity_check["calculated_velocity"]
    == velocity_check["transaction_velocity_1h"]
)

print("TRANSACTION VELOCITY VALIDATION")
print("-" * 45)

print("Total transactions:", len(velocity_check))
print("Matching velocity values:",
      velocity_check["velocity_match"].sum())
print("Non-matching velocity values:",
      (~velocity_check["velocity_match"]).sum())

print("\nRECORDED VELOCITY DISTRIBUTION")
display(
    transactions["transaction_velocity_1h"]
    .value_counts()
    .sort_index()
)

print("\nSAMPLE VELOCITY MISMATCHES")
display(
    velocity_check.loc[
        ~velocity_check["velocity_match"],
        [
            "transaction_id",
            "account_id",
            "timestamp",
            "transaction_velocity_1h",
            "calculated_velocity"
        ]
    ].head(10)
)
```

##### Key findings:
- 98.90% of transactions have a recorded velocity of zero, meaning no earlier transaction from the same account was observed within the preceding hour.
- 1,386 transactions have at least one earlier transaction within the preceding hour.
- Only 53 transactions have a velocity of two or three.
- The recorded feature matches our calculations exactly, so no corrections are indicated by this check.

```python

# STEP 2.5.2: Handle missing values

# 1. Fill missing merchant names
transactions_clean["merchant_name"] = (
    transactions_clean["merchant_name"].fillna("Unknown")
)

# 2. Create missing-device indicator
transactions_clean["device_info_missing"] = (
    transactions_clean["device_id"].isna().astype(int)
)

# 3. Preserve missing device IDs and new-device flags
# These values will be handled during model preprocessing.

# Verify cleaning results
print("FINLORA — MISSING VALUE CLEANING")
print("-" * 45)

print(
    "Missing merchant names:",
    transactions_clean["merchant_name"].isna().sum()
)

print(
    "Unknown merchant names:",
    (transactions_clean["merchant_name"] == "Unknown").sum()
)

print(
    "Missing device IDs:",
    transactions_clean["device_id"].isna().sum()
)

print(
    "Missing new-device flags:",
    transactions_clean["is_new_device"].isna().sum()
)

print("\nDEVICE INFORMATION INDICATOR")

display(
    transactions_clean["device_info_missing"]
    .value_counts()
)

print("\nUpdated transaction shape:",
      transactions_clean.shape)
```

```python

# STEP 2.5.3: Flag negative monetary values

transactions_clean["is_negative_amount"] = (
    transactions_clean["amount"] < 0
).astype(int)

transactions_clean["is_negative_baseline"] = (
    transactions_clean["avg_transaction_amount_30d"] < 0
).astype(int)

# Verify results
print("FINLORA — NEGATIVE MONETARY VALUE FLAGS")
print("-" * 45)

print(
    "Negative transaction amounts:",
    transactions_clean["is_negative_amount"].sum()
)

print(
    "Negative spending baselines:",
    transactions_clean["is_negative_baseline"].sum()
)

print(
    "Transactions with both negative:",
    (
        (transactions_clean["is_negative_amount"] == 1)
        & (transactions_clean["is_negative_baseline"] == 1)
    ).sum()
)

print(
    "\nUpdated transaction shape:",
    transactions_clean.shape
)
```

```python

# STEP 2.5.4: Flag account chronology inconsistencies

# Join account creation dates onto the transaction working copy
transactions_clean = transactions_clean.merge(
    accounts_clean[
        ["account_id", "account_created_date"]
    ],
    on="account_id",
    how="left",
    validate="many_to_one"
)

# Flag transactions dated before account creation
transactions_clean["account_date_inconsistent"] = (
    transactions_clean["timestamp"]
    < transactions_clean["account_created_date"]
).astype(int)

# Verify results
print("FINLORA — ACCOUNT CHRONOLOGY CLEANING")
print("-" * 45)

print(
    "Chronology inconsistencies:",
    transactions_clean["account_date_inconsistent"].sum()
)

print(
    "Consistent transactions:",
    (transactions_clean["account_date_inconsistent"] == 0).sum()
)

print(
    "Missing account creation dates:",
    transactions_clean["account_created_date"].isna().sum()
)

print(
    "\nUpdated transaction shape:",
    transactions_clean.shape
)

print("\nCHRONOLOGY FLAG DISTRIBUTION")

display(
    transactions_clean["account_date_inconsistent"]
    .value_counts()
)
```

```python

# STEP 2.5.5: Detect amount outliers by currency

# Work on a temporary dataset
outlier_check = transactions_clean[
    ["transaction_id", "currency", "amount", "is_fraud"]
].copy()

# Calculate IQR boundaries separately for each currency
currency_stats = outlier_check.groupby("currency")["amount"].agg(
    Q1=lambda x: x.quantile(0.25),
    Q3=lambda x: x.quantile(0.75)
)

currency_stats["IQR"] = (
    currency_stats["Q3"] - currency_stats["Q1"]
)

currency_stats["lower_bound"] = (
    currency_stats["Q1"] - 1.5 * currency_stats["IQR"]
)

currency_stats["upper_bound"] = (
    currency_stats["Q3"] + 1.5 * currency_stats["IQR"]
)

# Attach the currency-specific boundaries
outlier_check = outlier_check.join(
    currency_stats[["lower_bound", "upper_bound"]],
    on="currency"
)

# Identify statistical outliers
outlier_check["amount_outlier"] = (
    (outlier_check["amount"] < outlier_check["lower_bound"])
    | (outlier_check["amount"] > outlier_check["upper_bound"])
)

print("FINLORA — TRANSACTION AMOUNT OUTLIER INSPECTION")
print("-" * 50)

print("\nIQR BOUNDARIES BY CURRENCY")
display(currency_stats.round(2))

print("\nOUTLIER SUMMARY BY CURRENCY")

outlier_summary = outlier_check.groupby("currency").agg(
    total_transactions=("transaction_id", "count"),
    outliers=("amount_outlier", "sum")
)

outlier_summary["outlier_percentage"] = (
    outlier_summary["outliers"]
    / outlier_summary["total_transactions"] * 100
).round(2)

display(outlier_summary)

print("\nFRAUD RATE BY OUTLIER STATUS")

fraud_summary = outlier_check.groupby("amount_outlier").agg(
    transactions=("transaction_id", "count"),
    fraud_cases=("is_fraud", "sum"),
    fraud_rate=("is_fraud", "mean")
)

fraud_summary["fraud_rate"] = (
    fraud_summary["fraud_rate"] * 100
).round(2)

display(fraud_summary)

print("\nTOTAL STATISTICAL OUTLIERS:",
      outlier_check["amount_outlier"].sum())
```

#### Key findings:
- 20,979 transactions (16.65%) are statistical amount outliers.
- Non-outlier transactions have a 2.04% fraud rate.
- Outlier transactions have a 5.76% fraud rate.
- The fraud rate among outliers is approximately 2.8 times higher than among non-outliers.
- Outliers account for 1,209 of the 3,352 fraud cases, approximately 36.1% of all recorded fraud.
